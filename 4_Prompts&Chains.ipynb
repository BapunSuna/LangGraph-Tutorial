## **Prompts in Langchain**

```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI

from dotenv import load_dotenv

if os.environ['AICREDITS_API_KEY']:
    print("AICredits API Key is set.")
else:
    raise ValueError("AICredits API Key is not set.")
```

```python
import langchain_core
from langchain_core.messages import HumanMessage
llm = ChatOpenAI(
    model="openai/gpt-5-nano",
    base_url="https://api.aicredits.in/v1",
    api_key=os.environ["AICREDITS_API_KEY"],
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

my_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant that helps to create and curate posts for a social media platform."),
        ("human", "I want to create a post about {topic}")
    ]
)
```

## **Chain with Prompts**

```python
chain = my_prompt | llm
```

```python
chain.invoke({"topic": "The importance of data privacay in digital age"})
```
